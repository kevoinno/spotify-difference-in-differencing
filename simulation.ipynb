```python
import pandas as pd
import numpy as np

# Parameters
total_users = 10000
base_rate_suburb = 0.1
base_rate_influence = 0.09  # Slightly different to create parallel but not identical trends
trend_factor = 0.001  # Small increasing trend for both groups
marketing_effect = 0.05
max_listening_hours = 3
marketing_start_date = '2024-03-16'

# Generate date range for March 2024
dates = pd.date_range(start='2024-03-01', end='2024-03-31')

# Initialize user data
users = pd.DataFrame({
    'user_id': range(1, total_users + 1),
    'group': ['Suburb Talks'] * (total_users // 2) + ['Under the Influence'] * (total_users // 2),
})

# Simulate daily listening hours
listening_data = []

for date in dates:
    for index, row in users.iterrows():
        user_id = row['user_id']
        group = row['group']
        
        # Determine probability of listening with a slight upward trend
        days_passed = (date - dates[0]).days
        if group == 'Suburb Talks':
            listening_probability = base_rate_suburb + (trend_factor * days_passed)
        else:
            listening_probability = base_rate_influence + (trend_factor * days_passed)
        
        # Add marketing effect after the start date for Suburb Talks
        if group == 'Suburb Talks' and date >= pd.Timestamp(marketing_start_date):
            listening_probability += marketing_effect
        
        # Simulate whether the user listens and for how long
        listens = np.random.rand() < listening_probability
        listening_hours = np.random.randint(1, max_listening_hours + 1) if listens else 0
        
        listening_data.append({
            'user_id': user_id,
            'group': group,
            'date': date,
            'listening_hours': listening_hours
        })

# Create DataFrame from simulated data
listening_df = pd.DataFrame(listening_data)

# Summarize data
summary = listening_df.groupby(['group', 'date']).agg(total_listening_hours=('listening_hours', 'sum')).reset_index()
```

```python
summary.to_csv('summary.csv', index = False)
```
